1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

# Встановлення символьної змінної
x = sp.symbols('x')

# Функція f(x)
f = 10 * x - 5

# Обчислення похідної
df = sp.diff(f, x)

# Підготовка даних для візуалізації
x_vals = np.linspace(0, 10, 100)
f_vals = [f.subs(x, val) for val in x_vals]
df_vals = [df.subs(x, val) for val in x_vals]

# Побудова графіків
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.plot(x_vals, f_vals, label='f(x) = 10x - 5')
plt.title('Функція оцінки')
plt.xlabel('Години вивчення')
plt.ylabel('Оцінка')
plt.grid(True)
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(x_vals, df_vals, label="f'(x) = 10")
plt.title('Похідна функції оцінки')
plt.xlabel('Години вивчення')
plt.ylabel('Зміна оцінки')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
# Встановлення символьної змінної
x = sp.symbols('x')

# Визначення функцій і похідних
f_sin = sp.sin(x)
f_ln = sp.ln(x)

df_sin = sp.diff(f_sin, x)
df_ln = sp.diff(f_ln, x)

# Генерація даних для візуалізації
x_vals = np.linspace(0.1, 10, 400)
sin_vals = np.sin(x_vals)
sin_deriv_vals = np.cos(x_vals)
ln_vals = np.log(x_vals)
ln_deriv_vals = 1 / x_vals

# Побудова графіків
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.plot(x_vals, sin_vals, label='sin(x)')
plt.plot(x_vals, sin_deriv_vals, label="sin'(x) = cos(x)")
plt.title('Функція sin(x) та її похідна')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(x_vals, ln_vals, label='ln(x)')
plt.plot(x_vals, ln_deriv_vals, label="ln'(x) = 1/x")
plt.title('Функція ln(x) та її похідна')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

```


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
# Встановлення символьної змінної
x = sp.symbols('x')

# Функція витрат компанії
f = x**2 - 5*x + 1

# Обчислення похідної
df = sp.diff(f, x)

# Підготовка даних для візуалізації
x_vals = np.linspace(0, 100, 400)
f_vals = [f.subs(x, val) for val in x_vals]
df_vals = [df.subs(x, val) for val in x_vals]

# Побудова графіків
plt.figure(figsize=(10, 5))
plt.plot(x_vals, f_vals, label='f(x) = $x^2 - 5x + 1$')
plt.plot(x_vals, df_vals, label="f'(x) = $2x - 5$", linestyle='--')
plt.title('Графік функції витрат та її похідної')
plt.xlabel('Кількість вироблених одиниць (x)')
plt.ylabel('Вартість')
plt.legend()
plt.grid(True)
plt.show()
```

Функція витрат показує, як загальні витрати змінюються в залежності від кількості вироблених одиниць. Похідна функції витрат, або граничні витрати, показує швидкість зміни загальних витрат при виробництві додаткової одиниці продукції.

На графіку видно, що мінімальні загальні витрати досягаються приблизно при кількості одиниць, де похідна перетинає вісь абсцис (x = 2.5). Це означає, що до цього моменту витрати зменшуються, а після — збільшуються.
