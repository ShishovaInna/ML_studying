1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

```python
x = sp.symbols('x')
f = 10 * x - 5
derivative_f = sp.diff(f, x)
print("Похідна лінійної функції 10x - 5:", derivative_f)
```

```python
x = np.linspace(0, 10, 100)
f_values = 10 * x - 5
# Створення графіку
plt.figure(figsize=(10, 5))
plt.plot(x, f_values, label="f(x) = 10x-5", color='blue')
plt.plot(x, np.full_like(x, derivative_f), label="f'(x) = 10", color='green')

# Додавання легенди, міток та заголовка
plt.title("Графік функції та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```

```python
Чим більше годин навчаємось, тим вище оцінка. Похідка це константа, тому швидкість зростання оцінки не змінюється.
```

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
x = sp.symbols('x')

fs = sp.sin(x)
fl = sp.log(x)

df_s = sp.diff(fs, x)
df_l = sp.diff(fl, x)

print("Похідна функції f(x)=sin(x):", df_s)
print("Похідна функції f(x)=ln(x):", df_l)
```

```python
x = np.linspace(0, 10, 100)
fs = np.sin(x)
df_s = np.cos(x)
# Створення графіку
plt.figure(figsize=(10, 5))
plt.plot(x, fs, label="f(x)=sin(x)", color='blue')
plt.plot(x, df_s, label="f'(x)=cos(x)", color='green')

# Додавання легенди, міток та заголовка
plt.title("Графік функції та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```

```python
x = np.linspace(0.1, 10, 100)
fl = np.log(x)
df_l = 1 / x
# Створення графіку
plt.figure(figsize=(10, 5))
plt.plot(x, fl, label="f(x)=ln(x)", color='orange')
plt.plot(x, df_l, label="f'(x)=1/x", color='red')

# Додавання легенди, міток та заголовка
plt.title("Графік функції та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
x = sp.symbols('x')
f = x**2 - 5*x + 1
derivative_f = sp.diff(f, x)
print("Похідна лінійної функції x^2−5x+1:", derivative_f)
```

```python
x = np.linspace(0, 100, 100)
f = x**2 - 5*x + 1
df = 2*x - 5
# Створення графіку
plt.figure(figsize=(10, 5))
plt.plot(x, f, label="f(x)=x^2−5x+1", color='blue')
plt.plot(x, df, label="f'(x)=2*x - 5", color='green')

# Додавання легенди, міток та заголовка
plt.title("Графік функції та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```
