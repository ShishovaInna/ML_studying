```python
A = np.array(
    [[0.18, 0.53],
     [0.75, 0.47],
     [0.35, 0.21]])

B = np.array(
    [[0.17, 0.58, 0.75],
     [0.38, 0.11, 0.15]])

C = np.array([[0], [1]])

D = np.array([[0,1], [1,0]])

display(A, B, C, D)
```

```python
result_matrix = np.dot(B, A)
result_matrix
```

```python
matrix_C = np.dot(result_matrix, C)
matrix_C, matrix_C.shape
```

```python
matrix_D = np.dot(result_matrix, D)
matrix_D, matrix_D.shape
```

По вектору v1 транспонування не спрацювало мабудь тому, що вектор-рядок був одновимірним і Python не зміг його перевернути.
